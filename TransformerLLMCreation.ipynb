# Project Overview: Custom 3-Layer Decoder Transformer for Indian Cultural Knowledge

### Introduction
This project aims to design, implement, and evaluate a specialized **Decoder-only Transformer model** with a 3-layer architecture. Unlike general-purpose models, this Transformer is specifically trained on a synthetically generated dataset focused on **Indian culture, geography, and landmarks**.

### The 3-Layer Architecture Explained
To make this accessible to all stakeholders, we break down our '3-layer' approach as follows:
1.  **Input Layer (Contextualization)**: Where the model receives and encodes the text sequences into numbers.
2.  **Hidden Layer (Processing/Attention)**: The core 'brain' where the model uses self-attention to understand the relationships between different Indian entities (e.g., linking 'Karnataka' to 'Bangalore').
3.  **Output Layer (Prediction)**: Where the model generates the most probable next word in the sequence.

### Why Indian Cultural Data?
By using a targeted synthetic dataset, we ensure high-quality 'ground truth' information. This allows us to strictly evaluate the model's ability to learn specific patterns without the noise found in general web-scraped data.

## Synthetic Data Generation for Indian Context

### Subtask:
Programmatically generate a dataset covering Indian states, capitals, landmarks, and festivals to serve as the ground truth.


**Reasoning**:
I will start by creating the raw data structure and the sentence generation logic using Python.



```python
import random

# Data structure representing Indian cultural and geographical facts
india_data = [
    {"state": "Karnataka", "capital": "Bangalore", "landmark": "Mysore Palace", "festival": "Mysore Dasara"},
    {"state": "Maharashtra", "capital": "Mumbai", "landmark": "Gateway of India", "festival": "Ganesh Chaturthi"},
    {"state": "Tamil Nadu", "capital": "Chennai", "landmark": "Meenakshi Temple", "festival": "Pongal"},
    {"state": "West Bengal", "capital": "Kolkata", "landmark": "Victoria Memorial", "festival": "Durga Puja"},
    {"state": "Rajasthan", "capital": "Jaipur", "landmark": "Hawa Mahal", "festival": "Pushkar Mela"},
    {"state": "Kerala", "capital": "Thiruvananthapuram", "landmark": "Backwaters", "festival": "Onam"},
    {"state": "Punjab", "capital": "Chandigarh", "landmark": "Golden Temple", "festival": "Baisakhi"},
    {"state": "Uttar Pradesh", "capital": "Lucknow", "landmark": "Taj Mahal", "festival": "Diwali"},
    {"state": "Gujarat", "capital": "Gandhinagar", "landmark": "Statue of Unity", "festival": "Navratri"},
    {"state": "Telangana", "capital": "Hyderabad", "landmark": "Charminar", "festival": "Bonalu"}
]

def generate_sentences(data):
    sentences = []
    for item in data:
        s1 = f"The capital of {item['state']} is {item['capital']}."
        s2 = f"{item['state']} is known for the {item['landmark']}."
        s3 = f"The most famous festival in {item['state']} is {item['festival']}."
        s4 = f"{item['capital']} is the capital city of {item['state']}."
        sentences.extend([s1, s2, s3, s4])
    return sentences

# Generate corpus
raw_corpus = generate_sentences(india_data)
# Shuffle to ensure varied context during training
random.shuffle(raw_corpus)

print(f"Generated {len(raw_corpus)} ground truth sentences.")
print("Sample sentences:")
for s in raw_corpus[:5]:
    print(f"- {s}")
```

## Data Tokenization and Formatting

### Subtask:
Transform the text data into numerical sequences suitable for model training.


**Reasoning**:
I will begin by building the character-level vocabulary and the mapping dictionaries, then process the corpus into a numerical format.



```python
import torch

# Create a set of all characters in the corpus
chars = sorted(list(set(''.join(raw_corpus))))
vocab_size = len(chars)

# Create mapping from character to index and vice versa
char_to_int = {ch: i for i, ch in enumerate(chars)}
int_to_char = {i: ch for i, ch in enumerate(chars)}

def encode(s):
    return [char_to_int[c] for c in s]

def decode(l):
    return ''.join([int_to_char[i] for i in l])

# Tokenize the entire corpus into a single sequence
data = torch.tensor(encode(' '.join(raw_corpus)), dtype=torch.long)

print(f"Vocabulary size: {vocab_size}")
print(f"Characters found: {''.join(chars)}")
print(f"Total tokens in corpus: {len(data)}")
print(f"Sample encoded data (first 20 tokens): {data[:20].tolist()}")
```

### The Importance of Tokenization and Numerical Formatting

To build a Transformer model, we must first translate human language into a language that a computer's processor can understand: **Numbers**.

#### Why Numbers?
Neural networks are essentially massive mathematical functions. They perform operations like addition and multiplication. Since they cannot 'read' letters directly, we map every unique character (or word) to a unique integer.

#### Business & Technical Impact:
*   **Vocabulary Size**: By using character-level tokenization, we keep the 'Vocabulary Size' small (44 unique characters in our case). This makes the model computationally efficient and lightweight.
*   **Data Quality**: Our 'Ground Truth' corpus was processed into a single sequence of 1770 tokens. This precise mapping ensures that when the model generates text later, it pulls from a controlled set of Indian cultural facts, minimizing the initial risk of gibberish output.
*   **Encoding/Decoding**: We have established a two-way street. The `encode` function converts our Indian cultural facts into tensors for the 'brain', and the `decode` function allows us to read the model's numerical predictions back as human text.

## Building the 3-Layer Decoder Transformer

### Subtask:
Implement a custom Decoder-only Transformer architecture with three layers of self-attention.


**Reasoning**:
I will define the core components of the Transformer: the Self-Attention head, Multi-Head Attention, Feed-Forward network, and the Transformer Block, then assemble them into a 3-layer Decoder model.



```python
import torch.nn as nn
from torch.nn import functional as F

# Hyperparameters
batch_size = 16
block_size = 32
n_embd = 64
n_head = 4
n_layer = 3
dropout = 0.1
device = 'cuda' if torch.cuda.is_available() else 'cpu'

class Head(nn.Module):
    """ one head of self-attention """
    def __init__(self, head_size):
        super().__init__()
        self.key = nn.Linear(n_embd, head_size, bias=False)
        self.query = nn.Linear(n_embd, head_size, bias=False)
        self.value = nn.Linear(n_embd, head_size, bias=False)
        self.register_buffer('tril', torch.tril(torch.ones(block_size, block_size)))
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        B,T,C = x.shape
        k = self.key(x)
        q = self.query(x)
        wei = q @ k.transpose(-2,-1) * k.shape[-1]**-0.5
        wei = wei.masked_fill(self.tril[:T, :T] == 0, float('-inf'))
        wei = F.softmax(wei, dim=-1)
        wei = self.dropout(wei)
        v = self.value(x)
        out = wei @ v
        return out

class MultiHeadAttention(nn.Module):
    def __init__(self, num_heads, head_size):
        super().__init__()
        self.heads = nn.ModuleList([Head(head_size) for _ in range(num_heads)])
        self.proj = nn.Linear(head_size * num_heads, n_embd)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        out = torch.cat([h(x) for h in self.heads], dim=-1)
        out = self.dropout(self.proj(out))
        return out

class FeedForward(nn.Module):
    def __init__(self, n_embd):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(n_embd, 4 * n_embd),
            nn.ReLU(),
            nn.Linear(4 * n_embd, n_embd),
            nn.Dropout(dropout),
        )

    def forward(self, x):
        return self.net(x)

class Block(nn.Module):
    def __init__(self, n_embd, n_head):
        super().__init__()
        head_size = n_embd // n_head
        self.sa = MultiHeadAttention(n_head, head_size)
        self.ffwd = FeedForward(n_embd)
        self.ln1 = nn.LayerNorm(n_embd)
        self.ln2 = nn.LayerNorm(n_embd)

    def forward(self, x):
        x = x + self.sa(self.ln1(x))
        x = x + self.ffwd(self.ln2(x))
        return x

class DecoderTransformer(nn.Module):
    def __init__(self):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, n_embd)
        self.position_embedding_table = nn.Embedding(block_size, n_embd)
        self.blocks = nn.Sequential(*[Block(n_embd, n_head=n_head) for _ in range(n_layer)])
        self.ln_f = nn.LayerNorm(n_embd)
        self.lm_head = nn.Linear(n_embd, vocab_size)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        tok_emb = self.token_embedding_table(idx)
        pos_emb = self.position_embedding_table(torch.arange(T, device=device))
        x = tok_emb + pos_emb
        x = self.blocks(x)
        x = self.ln_f(x)
        logits = self.lm_head(x)

        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)

        return logits, loss

model = DecoderTransformer().to(device)
print(f"3-Layer Decoder Transformer initialized on {device}.")
```

## Model Training

### Subtask:
Establish a training loop to optimize the 3-layer Decoder on the Indian cultural dataset. Explain the process of 'weight optimization' for different audiences.


**Reasoning**:
I will start by defining the batching and loss estimation functions, then initialize the optimizer and run the training loop.



```python
import torch

# Train/Val split
n = int(0.9 * len(data))
train_data = data[:n]
val_data = data[n:]

def get_batch(split):
    data_split = train_data if split == 'train' else val_data
    ix = torch.randint(len(data_split) - block_size, (batch_size,))
    x = torch.stack([data_split[i:i+block_size] for i in ix])
    y = torch.stack([data_split[i+1:i+block_size+1] for i in ix])
    x, y = x.to(device), y.to(device)
    return x, y

@torch.no_grad()
def estimate_loss():
    out = {}
    model.eval()
    for split in ['train', 'val']:
        losses = torch.zeros(200)
        for k in range(200):
            X, Y = get_batch(split)
            logits, loss = model(X, Y)
            losses[k] = loss.item()
        out[split] = losses.mean()
    model.train()
    return out

# Optimizer
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)

max_iters = 3000
eval_interval = 500

print("Starting training loop...")
for iter in range(max_iters):
    if iter % eval_interval == 0:
        losses = estimate_loss()
        print(f"step {iter}: train loss {losses['train']:.4f}, val loss {losses['val']:.4f}")

    xb, yb = get_batch('train')
    logits, loss = model(xb, yb)
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()

print("Training complete.")
```

### Understanding the Training Process: Weight Optimization

Training a Transformer is essentially a process of **Optimization**. We want the model to minimize the 'loss' (error) between its predictions and the actual ground truth sentences we generated.

#### For Technical Stakeholders
We used the **AdamW optimizer**, a sophisticated version of Stochastic Gradient Descent.
- **Backpropagation**: For every batch of data, the model calculates the gradient of the loss function with respect to every single parameter (weight) in the 3-layer architecture.
- **Learning Rate**: Set at `1e-3`, this determines the step size the optimizer takes toward the minimum error.
- **Validation Split**: By holding out 10% of the data, we monitor the 'Val Loss'. If the training loss drops while val loss rises, we know the model is over-memorizing (overfitting).

#### For Business Stakeholders
Think of the training process like a student practicing for a quiz on Indian Geography:
- **The Textbook**: Our synthetic dataset of 40 facts.
- **The Practice (Training Loop)**: The student reads the facts 3,000 times (iterations).
- **The Correction (Weights)**: Every time the student gets a fact wrong (e.g., saying the capital of Karnataka is Mumbai), we correct them. The student 'updates their brain' (weights) so they are less likely to make that mistake again.
- **Goal**: By the end of the 3,000 steps, the student should be able to complete a sentence like 'The capital of Karnataka is...' with high confidence and accuracy.

## Prompting and Inference

### Subtask:
Provide a testing function where we give a prompt about India. The model will generate text based on its learned knowledge of states and landmarks.


**Reasoning**:
I will implement a generation function that takes text prompts, encodes them, and uses the trained model to predict the subsequent characters based on the learned Indian cultural data.



```python
def generate_text(model, prompt, max_new_tokens=50):
    model.eval()
    # Encode the prompt
    context = torch.tensor(encode(prompt), dtype=torch.long, device=device).unsqueeze(0)

    # Generate tokens
    for _ in range(max_new_tokens):
        # Crop context to block_size
        context_cond = context[:, -block_size:]
        # Get predictions
        logits, _ = model(context_cond)
        # Focus only on the last time step
        logits = logits[:, -1, :]
        # Apply softmax to get probabilities
        probs = F.softmax(logits, dim=-1)
        # Sample from the distribution
        idx_next = torch.multinomial(probs, num_samples=1)
        # Append to the sequence
        context = torch.cat((context, idx_next), dim=1)

    return decode(context[0].tolist())

# Test prompts
test_prompts = [
    "The capital of Maharashtra is",
    "West Bengal is known for the",
    "The most famous festival in Kerala is",
    "Jaipur is the capital city of"
]

print("--- Model Inference Results ---\n")
for prompt in test_prompts:
    generated = generate_text(model, prompt, max_new_tokens=40)
    print(f"Prompt: {prompt}")
    print(f"Generated: {generated}\n")
```

## Technical Evaluation (Perplexity and Accuracy)

### Subtask:
Calculate mathematical metrics to show how well the model predicts tokens and its accuracy against the ground truth synthetic data.


**Reasoning**:
I will write the code to calculate both perplexity and character-level accuracy on the validation set to quantify the model's predictive power.



```python
import math
import torch

@torch.no_grad()
def calculate_technical_metrics(model, split):
    model.eval()
    X, Y = get_batch(split)
    logits, loss = model(X, Y)

    # Perplexity is exp(cross_entropy_loss)
    perplexity = math.exp(loss.item())

    # Accuracy: Compare predicted character index with target index
    # Logits shape is (B*T, C), Y shape is (B*T)
    B, T, C = logits.shape if len(logits.shape)==3 else (logits.shape[0], 1, logits.shape[1])
    preds = torch.argmax(logits, dim=-1)
    correct = (preds == Y.view(-1)).float().sum()
    accuracy = correct / Y.numel()

    return perplexity, accuracy.item()

# Evaluate on Validation Set
val_perplexity, val_accuracy = calculate_technical_metrics(model, 'val')

print(f"--- Technical Evaluation Metrics (Validation Set) ---")
print(f"Perplexity: {val_perplexity:.4f}")
print(f"Character-Level Accuracy: {val_accuracy * 100:.2f}%")

print("\nInterpretation:")
print(f"- Perplexity of {val_perplexity:.2f} suggests the model is choosing between roughly {int(val_perplexity)} characters at each step.")
print(f"- Accuracy of {val_accuracy * 100:.2f}% reflects how often the most likely character matches the ground truth.")
```

## Business Evaluation: Hallucination and Quality Analysis

While technical metrics like Perplexity and Accuracy give us a mathematical baseline, they don't tell the full story of whether a model is 'useful' or 'dangerous' in a real-world scenario.

### What is Hallucination?
In generative AI, **hallucination** occurs when a model generates text that is grammatically correct and fluent but factually incorrect.
- **Example from our model**: It might generate *"The capital of Maharashtra is Ganesh Chaturthi"* or *"Jaipur is the capital city of West Bengal"*.
- **Why it happens**: The model is predicting the next *character* based on patterns. If the word 'capital' is often followed by 'Maharashtra' and also 'Ganesh Chaturthi' appears in similar contexts (festivals), the model might confuse these relationships if the training data is small or the layers aren't deep enough.

### Business Impact: Tourism and Cultural Applications
If this model were deployed as a 'Virtual Tour Guide for India':
1.  **Trust Erosion**: Providing the wrong capital for a state immediately destroys user trust.
2.  **Operational Risk**: If a user relies on the model to find a landmark (e.g., looking for the 'Victoria Memorial' in 'Jaipur' because the model said so), it leads to a poor customer experience.
3.  **Creative vs. Factual**: For a poetry bot, hallucination might be seen as 'creativity'. For an informational bot, it is a critical failure.

### Mitigating the Risk
To move from this prototype to a production-ready system, we would employ several strategies:
- **Retrieval-Augmented Generation (RAG)**: Instead of relying on the model's internal memory, we would have the model look up facts in a verified database of Indian landmarks.
- **Temperature Tuning**: Lowering the 'temperature' during inference makes the model more deterministic and less likely to sample 'random' (and potentially wrong) characters.
- **Constraint Satisfaction**: Hard-coding rules that ensure 'capital' tokens are only followed by known 'city' tokens.
